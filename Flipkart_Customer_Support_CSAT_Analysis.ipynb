## 3. ***Data Wrangling***

````python
# Missing values count

df.isnull().sum()
````

````python
# Missing values percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage
````

````python
# Duplicate values

df.duplicated().sum()
````

````python
# Removing duplicate rows

df = df.drop_duplicates()
````

````python
# Convert date columns into datetime format

df['order_date_time'] = pd.to_datetime(df['order_date_time'], errors='coerce')
df['Issue_reported at'] = pd.to_datetime(df['Issue_reported at'], errors='coerce')
df['issue_responded'] = pd.to_datetime(df['issue_responded'], errors='coerce')
df['Survey_response_Date'] = pd.to_datetime(df['Survey_response_Date'], errors='coerce')
````

````python
# Copy original dataset
df_clean = df.copy()

# Check initial shape
print("Original shape:", df_clean.shape)

# Remove duplicate rows
df_clean.drop_duplicates(inplace=True)

# Convert date columns to datetime format
date_columns = ['order_date_time', 'Issue_reported at', 'issue_responded', 'Survey_response_Date']

for col in date_columns:
    df_clean[col] = pd.to_datetime(df_clean[col], errors='coerce')

# Create response time column in minutes
df_clean['response_time_minutes'] = (
    df_clean['issue_responded'] - df_clean['Issue_reported at']
).dt.total_seconds() / 60

# Fill missing categorical values
categorical_cols = [
    'Customer Remarks', 'Customer_City', 'Product_category',
    'channel_name', 'category', 'Sub-category',
    'Agent_name', 'Supervisor', 'Manager',
    'Tenure Bucket', 'Agent Shift'
]

for col in categorical_cols:
    df_clean[col] = df_clean[col].fillna('Unknown')

# Fill missing numerical values
numerical_cols = ['Item_price', 'connected_handling_time', 'response_time_minutes']

for col in numerical_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Create CSAT category column
df_clean['CSAT_Category'] = df_clean['CSAT Score'].apply(
    lambda x: 'Satisfied' if x >= 4 else 'Dissatisfied'
)

# Final check
print("Cleaned shape:", df_clean.shape)
df_clean.head()
````

````python
# Checking missing values after cleaning
df_clean.isnull().sum()
````

The dataset was prepared for analysis by removing duplicate records, converting date columns into datetime format, creating a response_time_minutes column, and handling missing values. Missing categorical values were filled with "Unknown", while missing numerical values were filled using median values. A new CSAT_Category column was created to classify customers as satisfied or dissatisfied based on CSAT Score.
